```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [4, 67, 5, 45, 44]
y = [56, 65, 34, 12, 0]
```

```python
plt.scatter(x, y)
plt.show()
```

```python
x = np.random.random(100)
y = np.random.random(100)
colors= np.random.random(100)
sizes = np.random.random(100)*500
plt.scatter(x, y, c=colors, s=sizes, alpha=0.8, cmap="hot")
plt.show()
```

```python
plt.scatter(x, y, label="red", color='red', marker='p')
plt.scatter(y, x, label="blue", color='blue', marker='1', s=sizes)
plt.legend()
plt.show()
```
