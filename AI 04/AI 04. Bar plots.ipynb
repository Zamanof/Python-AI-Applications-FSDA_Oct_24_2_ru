```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
labels = ["Python", "C#", "JavaScript", "Java"]

percents = [35, 300, 20, 100]
```

```python
plt.bar(labels, percents, color=["red", "blue", "gray", "aqua"])
plt.title("Programming Languages")
plt.xlabel("Languages")
plt.ylabel("Percents")
plt.show()
```

```python
plt.barh(labels, percents, color=["red", "blue", "gray", "aqua"])
plt.title("Programming Languages")
plt.xlabel("Languages")
plt.ylabel("Percents")
for index, value in enumerate(percents):
    plt.text(index, index+0.42, str(value))
plt.show()
```
