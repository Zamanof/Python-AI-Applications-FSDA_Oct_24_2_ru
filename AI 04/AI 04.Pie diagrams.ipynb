```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
labels = ["Python", "C#", "JavaScript", "Java"]

percents = [35, 30, 20, 100]
```

```python
plt.pie(percents, labels=labels, autopct="%1.2f%%")
plt.title("Programming Languages")
plt.show()
```

```python
exploded = [0, 0.1, 0, 0]
plt.pie(percents, labels=labels, autopct="%1.2f%%",
        explode=exploded, shadow=True,
        startangle=45)
plt.title("Programming Languages")
plt.show()
```
