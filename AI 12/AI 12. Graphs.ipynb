# Граф вычислений
$$
y = X^2 + 5X + 7
$$

1. Посчитать $X^2$
2. Посчитать $5X$
3. Сложить этот результат и добавить 7



```python
X = torch.tensor(2.0, requires_grad=True)
Y = X**2 + 5*X + 7
X, Y
```

```python
Y.grad_fn
```

$$

$$

$$

$$
\frac{dy}{dx} = 2*2 + 5 = 9
$$

```python
Y.backward()
print(X.grad)
```
