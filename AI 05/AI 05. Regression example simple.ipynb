```python
import numpy as np
import matplotlib.pyplot as plt
```

# Функция регрессии

$$y=f(x_1, x_2,...,x_n)+\sigma$$

- $f$ - функция зависимости
- $y$ - Целевая переменная
- $x_i$ - признаки
- $\sigma$ - ошибка (шум)

# Математическая формула

$$y=a+bx+\sigma$$

- $y$ - Целевая переменная
- $x$ - независимая переменная (фактор)
- $a$ - свободный член (intercept)
- $b$ - коофциент наклона (slope)
- $\sigma$ - случайная ошибка (разница между реальным и предсказанным значением)
$$\sigma = \hat{y} - y$$

```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([53, 59, 63, 70, 75])
```

```python
n = len(x)
```

$$y=a+bx+\sigma$$

$$
b=\frac{n\sum{xy} - \sum{x}\sum{y}}{n\sum{x^2} - (\sum{x})^2}
$$

$$
a=\frac{\sum{y} - b\sum{x}}{n}
$$

```python
b = ((n * np.sum(x*y)) - np.sum(x)*np.sum(y))/(n*np.sum(x**2)-(np.sum(x))**2)

a = (np.sum(y) - b * np.sum(x))/n
```

```python
print(f"Уравнение y = {a:.2f} + {b:.2f} * x")
```

```python
plt.scatter(x, y, color='red', label='Данные', zorder=2)
plt.plot(x, y, color='blue', label='Регрессия', zorder=1)
plt.xlabel('Часы подготовки')
plt.ylabel('Оценка')
plt.title('Простая линейная регрессия')
plt.legend()
plt.show()
```

```python
while True:
    try:
        x_new = float(input("Введите часы подготовки: "))
        if x_new == 0:
            print("Exit")
            break
        y_new = a + b * x_new
        print(f'Предсказано y = {y_new:.2f}')
        plt.scatter(x, y, color='red', label='Данные', zorder=2)
        plt.plot(x, y, color='blue', label='Регрессия', zorder=1)
        plt.scatter(x_new, y_new, color='aqua', label='Наше предсказывание', zorder=2, s=100)
        plt.xlabel('Часы подготовки')
        plt.ylabel('Оценка')
        plt.title(f'Предсказывание для {x_new} часов')
        plt.legend()
        plt.show()
    except ValueError:
        print("Пожайлуста введите число")
```
